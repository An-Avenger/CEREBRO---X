# Cerebro X - Cloud GPU Training Runner

This notebook extracts the uploaded Cerebro X codebase, sets up the environment, and runs the deep learning training scripts (Phase 4 and Phase 5) on the Kaggle GPU.

```python
import os
import shutil
import zipfile
import sys

# 1. Extract the codebase zip if it was uploaded as a dataset
# Change this path if your dataset is named differently
CODEBASE_ZIP = "/kaggle/input/cerebro-x-codebase/cerebro_x_codebase.zip"
WORKING_DIR = "/kaggle/working/cerebro-x"

if os.path.exists(CODEBASE_ZIP):
    print(f"Extracting codebase from {CODEBASE_ZIP}...")
    with zipfile.ZipFile(CODEBASE_ZIP, 'r') as zip_ref:
        zip_ref.extractall(WORKING_DIR)
    print("Extraction complete.")
else:
    print("Zip file not found. Ensure you uploaded it as a dataset and named it correctly.")

# 2. Add the extracted src to Python path
if os.path.exists(WORKING_DIR):
    sys.path.insert(0, f"{WORKING_DIR}/src")
    os.chdir(WORKING_DIR)
    print("Working directory set to:", os.getcwd())
else:
    print("Error: Working directory not created.")
```
