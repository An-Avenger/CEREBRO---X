```python
import os
import zipfile

# Look for the uploaded cerebro_x_code.zip anywhere in /kaggle/input/
zip_path = None
for root, dirs, files in os.walk('/kaggle/input/'):
    if 'cerebro_x_code.zip' in files:
        zip_path = os.path.join(root, 'cerebro_x_code.zip')
        break

if zip_path:
    print(f"Found code at {zip_path}, extracting...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall("./")
    print("Code extracted!")
else:
    print("ERROR: Could not find cerebro_x_code.zip. Did you upload it?")
```
